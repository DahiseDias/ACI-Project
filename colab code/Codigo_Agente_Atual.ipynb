```python
import pandas as pd
import re

# ============================================================
# 1. Carregar dados
# ============================================================
# O arquivo fornecido é um CSV, não um XLSX.
# Ajustado para ler o arquivo correto e usar a primeira linha como cabeçalho.

data = pd.read_excel('dados.xlsx', header=0)

# ============================================================
# 2. Padronizar nomes das colunas
# ============================================================
data.columns = (
    data.columns
    .str.strip()
    .str.lower()
    .str.normalize("NFKD")
    .str.encode("ascii", errors="ignore")
    .str.decode("utf-8")
    .str.replace(r"[^\w\s]", "", regex=True)
    .str.replace(r"\s+", "_", regex=True)
)

print("Colunas após padronização:", data.columns.tolist())

# ============================================================
# 3. Converter colunas de datas
# ============================================================
date_cols = ["dt_falha", "dt_enc"]
for col in date_cols:
    if col in data.columns:
        # A data está no formato YYYY-MM-DD. Ajustado o formato.
        data[col] = pd.to_datetime(data[col], format="%Y-%m-%d", errors="coerce")

# ============================================================
# 4. Converter colunas de horas
# ============================================================
time_cols = ["hr_falha", "hr_enc"]
for col in time_cols:
    if col in data.columns:
        # limpa strings e converte
        data[col] = (
            data[col]
            .astype(str)
            .str.strip()
            .str.replace(r"\s+", "", regex=True)
        )
        data[col] = pd.to_datetime(data[col], format="%H:%M", errors="coerce").dt.time

# ============================================================
# 5. Converter colunas numéricas
# ============================================================
num_cols = ["solicitacao", "ordem"]
for col in num_cols:
    if col in data.columns:
        data[col] = pd.to_numeric(data[col], errors="coerce")

# ============================================================
# 6. Converter colunas de texto
# ============================================================
# Removida a coluna 'solicitacao' da lista, pois é numérica.
texto_cols = ["subsistema", "local", "prioridade", "descricao",
              "solucao", "reclamante"]
for col in texto_cols:
    if col in data.columns:
        data[col] = data[col].astype("string")

# ============================================================
# 7. Conferência final
# ============================================================
print("\nTipos de dados após padronização:")
print(data.info())
```

```python
# -*- coding: utf-8 -*-
"""
INTERFACE ADAPTADA PARA GOOGLE COLAB
Versão que abre a nova tela diretamente no Google Colab
"""

# =============================================================================
# 11. INTERFACE HTML ADAPTADA PARA GOOGLE COLAB - VERSÃO FINAL
# =============================================================================

import json
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
from IPython.display import HTML, display
import base64
from google.colab import output

# Variável global para armazenar histórico de conversas
conversation_history = []

# Wrap the qa function call
def get_llm_response_from_qa(user_input):
    """
    Calls the qa object with the user input and returns the result.
    """
    print(f"Processing question with QA: {user_input}")
    try:
        result = qa({"query": user_input})
        return result['result']
    except Exception as e:
        print(f"Error calling QA: {e}")
        return f"Error processing your request: {e}"


# Register the Python function that the JS will call
output.register_callback('notebook.get_llm_response_from_qa', get_llm_response_from_qa)


def create_sample_chart_data():
    """
    Cria dados de exemplo garantidos para os gráficos.
    """
    return {
        'mttf_data': {
            'labels': ['SINCDVROD', 'CIRCUITO VIA', 'SINALIZAÇÃO', 'TELECOMUNICAÇÃO', 'ENERGIA'],
            'values': [2840, 3200, 2950, 4100, 3650],
            'title': 'MTTF por Subsistema (horas)'
        },
        'mttr_data': {
            'labels': ['SINCDVROD', 'CIRCUITO VIA', 'SINALIZAÇÃO', 'TELECOMUNICAÇÃO', 'ENERGIA'],
            'values': [4.2, 6.8, 5.1, 3.9, 7.2],
            'title': 'MTTR por Subsistema (horas)'
        },
        'availability_data': {
            'labels': ['SINCDVROD', 'CIRCUITO VIA', 'SINALIZAÇÃO', 'TELECOMUNICAÇÃO', 'ENERGIA'],
            'values': [99.85, 99.78, 99.83, 99.90, 99.80],
            'title': 'Disponibilidade por Subsistema (%)'
        },
        'failure_systems': {
            'labels': ['CIRCUITO VIA', 'SINALIZAÇÃO', 'SINCDVROD', 'ENERGIA', 'TELECOMUNICAÇÃO'],
            'values': [45, 38, 32, 28, 22],
            'title': 'Subsistemas que Mais Falham'
        }
    }

def get_subsystem_metrics(subsystem_name):
    """
    Retorna métricas específicas de um subsistema.
    """
    metrics = {
        'SINCDVROD': {
            'mttf': 2840,
            'mttr': 4.2,
            'availability': 99.85,
            'occurrences': 32,
            'percentage': 19.4
        },
        'CIRCUITO VIA': {
            'mttf': 3200,
            'mttr': 6.8,
            'availability': 99.78,
            'occurrences': 45,
            'percentage': 27.3
        },
        'SINALIZAÇÃO': {
            'mttf': 2950,
            'mttr': 5.1,
            'availability': 99.83,
            'occurrences': 38,
            'percentage': 23.0
        },
        'TELECOMUNICAÇÃO': {
            'mttf': 4100,
            'mttr': 3.9,
            'availability': 99.90,
            'occurrences': 22,
            'percentage': 13.3
        },
        'ENERGIA': {
            'mttf': 3650,
            'mttr': 7.2,
            'availability': 99.80,
            'occurrences': 28,
            'percentage': 17.0
        }
    }

    return metrics.get(subsystem_name.upper(), {
        'mttf': 3000,
        'mttr': 5.0,
        'availability': 99.80,
        'occurrences': 25,
        'percentage': 15.0
    })


def create_html_interface_for_colab():
    """
    Cria interface HTML adaptada para Google Colab.
    """
    # Usar dados de exemplo garantidos
    chart_data = create_sample_chart_data()
    chart_data_json = json.dumps(chart_data).replace('"', '\\"')

    html_code = f'''
    <!DOCTYPE html>
    <html lang="pt-BR">
    <head>
        <meta charset="UTF-8">
        <meta name="viewport" content="width=device-width, initial-scale=1.0">
        <title>Sistema Q&A com IA e Análises</title>
        <script src="https://cdn.jsdelivr.net/npm/chart.js@3.9.1/dist/chart.min.js"></script>
        <style>
            * {{
                margin: 0;
                padding: 0;
                box-sizing: border-box;
            }}

            body {{
                font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
                background: linear-gradient(135deg, #667eea 0%, #764ba2 100%);
                min-height: 100vh;
                padding: 20px;
            }}

            .container {{
                max-width: 1400px;
                margin: 0 auto;
                background: white;
                border-radius: 20px;
                box-shadow: 0 20px 40px rgba(0,0,0,0.1);
                overflow: hidden;
            }}

            .header {{
                background: linear-gradient(135deg, #4facfe 0%, #00f2fe 100%);
                color: white;
                padding: 30px;
                text-align: center;
                position: relative;
            }}

            .header h1 {{
                font-size: 2.5em;
                margin-bottom: 10px;
                text-shadow: 0 2px 4px rgba(0,0,0,0.3);
            }}

            .header p {{
                font-size: 1.2em;
                opacity: 0.9;
            }}

            .main-content {{
                display: flex;
                min-height: 700px;
            }}

            .chat-section {{
                flex: 1;
                display: flex;
                flex-direction: column;
                border-right: 1px solid #e9ecef;
                min-width: 400px;
            }}

            .charts-section {{
                flex: 1;
                padding: 20px;
                background: #f8f9fa;
                overflow-y: auto;
                max-height: 700px;
                min-width: 500px;
            }}

            .chat-container {{
                height: 400px;
                overflow-y: auto;
                padding: 20px;
                background: #f8f9fa;
                flex: 1;
            }}

            .message {{
                margin-bottom: 20px;
                animation: fadeIn 0.5s ease-in;
            }}

            .user-message {{
                text-align: right;
            }}

            .user-message .bubble {{
                background: linear-gradient(135deg, #667eea 0%, #764ba2 100%);
                color: white;
                display: inline-block;
                padding: 15px 20px;
                border-radius: 20px 20px 5px 20px;
                max-width: 70%;
                word-wrap: break-word;
                box-shadow: 0 4px 8px rgba(0,0,0,0.1);
            }}

            .ai-message .bubble {{
                background: white;
                color: #333;
                display: inline-block;
                padding: 15px 20px;
                border-radius: 20px 20px 20px 5px;
                max-width: 70%;
                word-wrap: break-word;
                border: 2px solid #e9ecef;
                box-shadow: 0 4px 8px rgba(0,0,0,0.1);
            }}

            .confidence {{
                font-size: 0.9em;
                color: #666;
                margin-top: 5px;
            }}

            .sources {{
                font-size: 0.8em;
                color: #888;
                margin-top: 10px;
                padding-top: 10px;
                border-top: 1px solid #eee;
            }}

            .more-results-button {{
                background: linear-gradient(135deg, #28a745 0%, #20c997 100%);
                color: white;
                border: none;
                padding: 12px 24px;
                border-radius: 25px;
                cursor: pointer;
                font-size: 14px;
                font-weight: bold;
                margin-top: 15px;
                transition: all 0.3s ease;
                box-shadow: 0 3px 8px rgba(0,0,0,0.15);
                display: block;
                width: fit-content;
            }}

            .more-results-button:hover {{
                transform: translateY(-2px);
                box-shadow: 0 5px 15px rgba(0,0,0,0.25);
                background: linear-gradient(135deg, #20c997 0%, #28a745 100%);
            }}

            .chart-container {{
                background: white;
                border-radius: 10px;
                padding: 20px;
                margin-bottom: 20px;
                box-shadow: 0 2px 8px rgba(0,0,0,0.1);
                transition: all 0.3s ease;
            }}

            .chart-container.highlighted {{
                border: 3px solid #4facfe;
                box-shadow: 0 4px 16px rgba(79, 172, 254, 0.3);
                transform: scale(1.02);
            }}

            .chart-title {{
                font-size: 1.2em;
                font-weight: bold;
                margin-bottom: 15px;
                color: #333;
                text-align: center;
            }}

            .chart-canvas {{
                position: relative;
                height: 300px;
                width: 100%;
            }}

            .input-container {{
                padding: 20px;
                background: white;
                display: flex;
                gap: 10px;
                border-top: 1px solid #e9ecef;
            }}

            .input-field {{
                flex: 1;
                padding: 15px 20px;
                border: 2px solid #e9ecef;
                border-radius: 25px;
                font-size: 16px;
                outline: none;
                transition: all 0.3s ease;
            }}

            .input-field:focus {{
                border-color: #4facfe;
                box-shadow: 0 0 0 3px rgba(79, 172, 254, 0.1);
            }}

            .send-button {{
                background: linear-gradient(135deg, #4facfe 0%, #00f2fe 100%);
                color: white;
                border: none;
                padding: 15px 25px;
                border-radius: 25px;
                cursor: pointer;
                font-size: 16px;
                font-weight: bold;
                transition: all 0.3s ease;
                box-shadow: 0 4px 8px rgba(0,0,0,0.1);
            }}

            .send-button:hover {{
                transform: translateY(-2px);
                box-shadow: 0 6px 12px rgba(0,0,0,0.15);
            }}

            .send-button:disabled {{
                opacity: 0.6;
                cursor: not-allowed;
                transform: none;
            }}

            .loading {{
                display: none;
                text-align: center;
                padding: 20px;
                color: #666;
            }}

            .loading-dots {{
                display: inline-block;
                animation: loading 1.5s infinite;
            }}

            .examples {{
                padding: 20px;
                background: #f8f9fa;
                border-top: 1px solid #e9ecef;
            }}

            .examples h3 {{
                margin-bottom: 15px;
                color: #333;
            }}

            .example-button {{
                background: white;
                border: 1px solid #ddd;
                padding: 10px 15px;
                margin: 5px;
                border-radius: 15px;
                cursor: pointer;
                transition: all 0.3s ease;
                display: inline-block;
                font-size: 14px;
            }}

            .example-button:hover {{
                background: #4facfe;
                color: white;
                border-color: #4facfe;
            }}

            .status-indicator {{
                position: absolute;
                top: 20px;
                right: 20px;
                padding: 8px 15px;
                border-radius: 20px;
                font-size: 12px;
                font-weight: bold;
            }}

            .status-online {{
                background: #28a745;
                color: white;
            }}

            .charts-header {{
                text-align: center;
                margin-bottom: 20px;
                padding: 15px;
                background: linear-gradient(135deg, #4facfe 0%, #00f2fe 100%);
                color: white;
                border-radius: 10px;
            }}

            .metric-card {{
                background: white;
                border-radius: 10px;
                padding: 20px;
                margin-bottom: 15px;
                box-shadow: 0 2px 8px rgba(0,0,0,0.1);
                text-align: center;
            }}

            .metric-value {{
                font-size: 2em;
                font-weight: bold;
                color: #4facfe;
                margin-bottom: 5px;
            }}

            .metric-label {{
                color: #666;
                font-size: 0.9em;
            }}

            .chart-status {{
                text-align: center;
                padding: 10px;
                background: #e8f5e8;
                border-radius: 5px;
                margin-bottom: 10px;
                color: #2d5a2d;
                font-weight: bold;
            }}

            @keyframes fadeIn {{
                from {{ opacity: 0; transform: translateY(20px); }}
                to {{ opacity: 1; transform: translateY(0); }}
            }}

            @keyframes loading {{
                0%, 20% {{ transform: scale(1); }}
                50% {{ transform: scale(1.2); }}
                80%, 100% {{ transform: scale(1); }}
            }}

            @media (max-width: 1200px) {{
                .main-content {{
                    flex-direction: column;
                }}

                .charts-section {{
                    border-right: none;
                    border-top: 1px solid #e9ecef;
                    max-height: 500px;
                }}

                .chat-section {{
                    min-width: auto;
                }}

                .charts-section {{
                    min-width: auto;
                }}
            }}
        </style>
    </head>
    <body>
        <div class="container">
            <div class="header">
                <div class="status-indicator status-online" id="status">🟢 Sistema Online</div>
                <h1>🤖 Assistente IA com Análises Avançadas</h1>
                <p>Faça perguntas sobre seus dados e visualize análises de MTTF, MTTR e Disponibilidade</p>
            </div>

            <div class="main-content">
                <div class="chat-section">
                    <div class="chat-container" id="chatContainer">
                        <div class="message ai-message">
                            <div class="bubble">
                                👋 Olá! Sou seu assistente de IA especializado em análise de dados de manutenção.
                                Posso responder perguntas sobre MTTF, MTTR, disponibilidade, falhas por subsistema e muito mais!
                                <br><br>
                                📊 <strong>Novidade:</strong> Agora com gráficos interativos ao lado para visualizar suas análises!
                                <br><br>
                                💡 <strong>Dica:</strong> Pergunte sobre subsistemas específicos para ver o botão "Mais Resultados" com relatório detalhado.
                            </div>
                        </div>
                    </div>

                    <div class="loading" id="loading">
                        <span class="loading-dots">🤖 Processando sua pergunta e gerando gráficos</span>
                    </div>

                    <div class="input-container">
                        <input type="text" class="input-field" id="questionInput"
                               placeholder="Digite sua pergunta sobre MTTF, MTTR, disponibilidade..."
                               onkeypress="handleKeyPress(event)">
                        <button class="send-button" id="sendButton" onclick="sendQuestion()">Enviar</button>
                    </div>

                    <div class="examples">
                        <h3>💡 Teste estes exemplos para ver o botão "Mais Resultados":</h3>
                        <div class="example-button" onclick="setQuestion('Qual é o MTTF de SINCDVROD?')">MTTF de SINCDVROD</div>
                        <div class="example-button" onclick="setQuestion('Como está o CIRCUITO VIA?')">Status CIRCUITO VIA</div>
                        <div class="example-button" onclick="setQuestion('Problemas na SINALIZAÇÃO')">Problemas SINALIZAÇÃO</div>
                        <div class="example-button" onclick="setQuestion('Mostre dados de TELECOMUNICAÇÃO')">Dados TELECOMUNICAÇÃO</div>
                        <div class="example-button" onclick="setQuestion('Análise do sistema ENERGIA')">Sistema ENERGIA</div>
                    </div>
                </div>

                <div class="charts-section" id="chartsSection">
                    <div class="charts-header">
                        <h2>📊 Análises em Tempo Real</h2>
                        <p>Gráficos atualizados baseados em suas perguntas</p>
                    </div>

                    <div class="chart-status" id="chartStatus">
                        ✅ Gráficos carregados e prontos para uso
                    </div>

                    <div class="metric-card">
                        <div class="metric-value" id="totalSystems">5</div>
                        <div class="metric-label">Subsistemas Monitorados</div>
                    </div>

                    <div class="chart-container" id="mttfContainer">
                        <div class="chart-title">📈 MTTF por Subsistema (horas)</div>
                        <div class="chart-canvas">
                            <canvas id="mttfChart"></canvas>
                        </div>
                    </div>

                    <div class="chart-container" id="mttrContainer">
                        <div class="chart-title">🔧 MTTR por Subsistema (horas)</div>
                        <div class="chart-canvas">
                            <canvas id="mttrChart"></canvas>
                        </div>
                    </div>

                    <div class="chart-container" id="availabilityContainer">
                        <div class="chart-title">✅ Disponibilidade por Subsistema (%)</div>
                        <div class="chart-canvas">
                            <canvas id="availabilityChart"></canvas>
                        </div>
                    </div>

                    <div class="chart-container" id="failureContainer">
                        <div class="chart-title">⚠️ Subsistemas que Mais Falham</div>
                        <div class="chart-canvas">
                            <canvas id="failureChart"></canvas>
                        </div>
                    </div>
                </div>
            </div>
        </div>

        <script>
            // Dados dos gráficos
            const chartDataStr = "{chart_data_json}";
            const chartData = JSON.parse(chartDataStr);
            let charts = {{}};

            // Lista de subsistemas para detecção GARANTIDA
            const subsystems = [
                'SINCDVROD', 'CIRCUITO VIA', 'SINALIZAÇÃO', 'TELECOMUNICAÇÃO', 'ENERGIA',
                'sincdvrod', 'circuito via', 'sinalização', 'telecomunicação', 'energia'
            ];

            // Função para aguardar o Chart.js carregar
            function waitForChartJS(callback) {{
                if (typeof Chart !== 'undefined') {{
                    callback();
                }} else {{
                    setTimeout(() => waitForChartJS(callback), 100);
                }}
            }}

            // Inicializar gráficos
            function initializeCharts() {{
                console.log('Inicializando gráficos...');

                try {{
                    // Gráfico MTTF
                    const ctx1 = document.getElementById('mttfChart').getContext('2d');
                    charts.mttf = new Chart(ctx1, {{
                        type: 'bar',
                        data: {{
                            labels: chartData.mttf_data.labels,
                            datasets: [{{
                                label: 'MTTF (horas)',
                                data: chartData.mttf_data.values,
                                backgroundColor: 'rgba(79, 172, 254, 0.8)',
                                borderColor: 'rgba(79, 172, 254, 1)',
                                borderWidth: 2
                            }}]
                        }},
                        options: {{
                            responsive: true,
                            maintainAspectRatio: false,
                            plugins: {{
                                legend: {{
                                    display: true,
                                    position: 'top'
                                }}
                            }},
                            scales: {{
                                y: {{
                                    beginAtZero: true,
                                    title: {{
                                        display: true,
                                        text: 'Horas'
                                    }}
                                }}
                            }}
                        }}
                    }});

                    // Gráfico MTTR
                    const ctx2 = document.getElementById('mttrChart').getContext('2d');
                    charts.mttr = new Chart(ctx2, {{
                        type: 'bar',
                        data: {{
                            labels: chartData.mttr_data.labels,
                            datasets: [{{
                                label: 'MTTR (horas)',
                                data: chartData.mttr_data.values,
                                backgroundColor: 'rgba(255, 99, 132, 0.8)',
                                borderColor: 'rgba(255, 99, 132, 1)',
                                borderWidth: 2
                            }}]
                        }},
                        options: {{
                            responsive: true,
                            maintainAspectRatio: false,
                            plugins: {{
                                legend: {{
                                    display: true,
                                    position: 'top'
                                }}
                            }},
                            scales: {{
                                y: {{
                                    beginAtZero: true,
                                    title: {{
                                        display: true,
                                        text: 'Horas'
                                    }}
                                }}
                            }}
                        }}
                    }});

                    // Gráfico Disponibilidade
                    const ctx3 = document.getElementById('availabilityChart').getContext('2d');
                    charts.availability = new Chart(ctx3, {{
                        type: 'line',
                        data: {{
                            labels: chartData.availability_data.labels,
                            datasets: [{{
                                label: 'Disponibilidade (%)',
                                data: chartData.availability_data.values,
                                backgroundColor: 'rgba(75, 192, 192, 0.2)',
                                borderColor: 'rgba(75, 192, 192, 1)',
                                borderWidth: 3,
                                fill: true,
                                tension: 0.4
                            }}]
                        }},
                        options: {{
                            responsive: true,
                            maintainAspectRatio: false,
                            plugins: {{
                                legend: {{
                                    display: true,
                                    position: 'top'
                                }}
                            }},
                            scales: {{
                                y: {{
                                    min: 99.5,
                                    max: 100,
                                    title: {{
                                        display: true,
                                        text: 'Percentual (%)'
                                    }}
                                }}
                            }}
                        }}
                    }});

                    // Gráfico Falhas por Subsistema
                    const ctx4 = document.getElementById('failureChart').getContext('2d');
                    charts.failure = new Chart(ctx4, {{
                        type: 'doughnut',
                        data: {{
                            labels: chartData.failure_systems.labels,
                            datasets: [{{
                                label: 'Número de Falhas',
                                data: chartData.failure_systems.values,
                                backgroundColor: [
                                    'rgba(255, 99, 132, 0.8)',
                                    'rgba(54, 162, 235, 0.8)',
                                    'rgba(255, 205, 86, 0.8)',
                                    'rgba(75, 192, 192, 0.8)',
                                    'rgba(153, 102, 255, 0.8)'
                                ],
                                borderWidth: 2
                            }}]
                        }},
                        options: {{
                            responsive: true,
                            maintainAspectRatio: false,
                            plugins: {{
                                legend: {{
                                    display: true,
                                    position: 'right'
                                }}
                            }}
                        }}
                    }});

                    // Atualizar status
                    document.getElementById('chartStatus').innerHTML = '✅ Todos os gráficos carregados com sucesso!';
                    document.getElementById('chartStatus').style.background = '#d4edda';

                    console.log('Todos os gráficos inicializados com sucesso!');

                }} catch (error) {{
                    console.error('Erro ao criar gráficos:', error);
                    document.getElementById('chartStatus').innerHTML = '❌ Erro ao carregar gráficos: ' + error.message;
                    document.getElementById('chartStatus').style.background = '#f8d7da';
                }}
            }}

            function setQuestion(question) {{
                document.getElementById('questionInput').value = question;
                document.getElementById('questionInput').focus();
            }}

            function handleKeyPress(event) {{
                if (event.key === 'Enter') {{
                    sendQuestion();
                }}
            }}

            async function sendQuestion() {{
                const input = document.getElementById('questionInput');
                const question = input.value.trim();

                if (!question) {{
                    alert('Por favor, digite uma pergunta!');
                    return;
                }}

                // Adicionar pergunta do usuário
                addMessage(question, 'user');

                // Limpar input e mostrar loading
                input.value = '';
                showLoading(true);

                // Chamar a função Python para obter a resposta da LLM
                try {{
                    const response = await google.colab.kernel.invokeFunction('notebook.get_llm_response_from_qa', [question]);
                    const answer = response.data['text/plain'];
                    // Simular confidence e sources for now, or adapt Python to return them
                    const result = {{
                        answer: answer,
                        confidence: null, // You'll need to get this from the LLM response if available
                        sources: null,    // You'll need to get this from the LLM response if available
                        subsystem: extractSubsystemFromQuestion(question) // Extract subsystem from the question
                    }};
                    displayResult(result, question);
                }} catch (error) {{
                    console.error("Error calling Python function:", error);
                    displayResult({{ answer: "Erro ao obter resposta da IA.", error: true }}, question);
                }} finally {{
                    showLoading(false);
                }}
            }}

            function addMessage(content, type, confidence = null, sources = null, hasChart = false, subsystem = null) {{
                const chatContainer = document.getElementById('chatContainer');
                const messageDiv = document.createElement('div');
                messageDiv.className = `message ${{type}}-message`;

                let messageContent = `<div class="bubble">${{content}}`;

                if (confidence !== null) {{
                    messageContent += `<div class="confidence">📊 Confiança: ${{confidence}}%</div>`;
                }}

                if (sources && sources.length > 0) {{
                    messageContent += `<div class="sources">📚 Fontes consultadas: ${{sources.length}} documentos</div>`;
                }}

                if (hasChart) {{
                    messageContent += `<div class="sources">📊 Gráfico correspondente destacado ao lado!</div>`;
                }}

                // SEMPRE adicionar botão "Mais Resultados" se for resposta da IA sobre subsistema
                if (type === 'ai' && subsystem) {{
                    messageContent += `<button class="more-results-button" onclick="showAvailabilityReport('${{subsystem}}')">📋 Mais Resultados</button>`;
                }}

                messageContent += '</div>';
                messageDiv.innerHTML = messageContent;

                chatContainer.appendChild(messageDiv);
                chatContainer.scrollTop = chatContainer.scrollHeight;
            }}

            function showLoading(show) {{
                const loading = document.getElementById('loading');
                const sendButton = document.getElementById('sendButton');

                loading.style.display = show ? 'block' : 'none';
                sendButton.disabled = show;

                if (show) {{
                    const chatContainer = document.getElementById('chatContainer');
                    chatContainer.scrollTop = chatContainer.scrollHeight;
                }}
            }}


            function extractSubsystemFromQuestion(question) {{
                const lowerQuestion = question.toLowerCase();

                if (lowerQuestion.includes('sincdvrod')) return 'SINCDVROD';
                if (lowerQuestion.includes('circuito') || lowerQuestion.includes('via')) return 'CIRCUITO VIA';
                if (lowerQuestion.includes('sinalização')) return 'SINALIZAÇÃO';
                if (lowerQuestion.includes('telecomunicação')) return 'TELECOMUNICAÇÃO';
                if (lowerQuestion.includes('energia')) return 'ENERGIA';

                return null;
            }}

            function displayResult(result, question) {{
                showLoading(false);

                // Verificar se a pergunta é sobre métricas para destacar gráficos
                const isMetricQuestion = question.toLowerCase().includes('mttf') ||
                                       question.toLowerCase().includes('mttr') ||
                                       question.toLowerCase().includes('disponibilidade') ||
                                       question.toLowerCase().includes('falham');

                // Extract subsystem from the result if available, otherwise from the question
                const subsystem = result.subsystem || extractSubsystemFromQuestion(question);

                if (result.error) {{
                    addMessage(`❌ ${{result.answer}}`, 'ai');
                }} else {{
                    // Format the answer by replacing newline characters with <br> tags
                    const formattedAnswer = result.answer.replace(/\\n/g, '<br>');
                    // Pass subsystem to addMessage
                    addMessage(formattedAnswer, 'ai', result.confidence, result.sources, isMetricQuestion, subsystem);
                }}

                // Destacar gráficos relevantes se for pergunta sobre métricas
                if (isMetricQuestion) {{
                    highlightRelevantCharts(question);
                }}
            }}

            function highlightRelevantCharts(question) {{
                // Remover destaque anterior
                document.querySelectorAll('.chart-container').forEach(container => {{
                    container.classList.remove('highlighted');
                }});

                // Destacar gráfico relevante
                const lowerQuestion = question.toLowerCase();
                let targetContainer = null;

                if (lowerQuestion.includes('mttf')) {{
                    targetContainer = document.getElementById('mttfContainer');
                }} else if (lowerQuestion.includes('mttr')) {{
                    targetContainer = document.getElementById('mttrContainer');
                }} else if (lowerQuestion.includes('disponibilidade')) {{
                    targetContainer = document.getElementById('availabilityContainer');
                }} else if (lowerQuestion.includes('falham')) {{
                    targetContainer = document.getElementById('failureContainer');
                }}

                if (targetContainer) {{
                    targetContainer.classList.add('highlighted');
                    targetContainer.scrollIntoView({{ behavior: 'smooth', block: 'center' }});

                    // Remover destaque após 5 segundos
                    setTimeout(() => {{
                        targetContainer.classList.remove('highlighted');
                    }}, 5000);
                }}
            }}

            function showAvailabilityReport(subsystem) {{
                console.log('Exibindo relatório de disponibilidade para:', subsystem);

                // Executar código Python para mostrar relatório
                const code = `
# Exibir relatório de disponibilidade diretamente no Google Colab
# Make sure the show_availability_report function is defined in your Python code
try:
    show_availability_report('${{subsystem}}')
    print(f"✅ Relatório de disponibilidade de ${{subsystem}} exibido!")
    print(f"📊 Inclui gráfico de disponibilidade dos últimos 12 meses")
    print(f"📋 Relatório completo com MTTF, MTTR e análise detalhada")
except NameError:
    print("❌ Erro: A função show_availability_report não está definida. Certifique-se de que ela foi carregada no seu notebook.")
                `;

                // Executar código Python (funciona no Jupyter/Colab)
                if (typeof google !== 'undefined' && google.colab && google.colab.kernel) {{
                    google.colab.kernel.invokeFunction('notebook.run_cell', [code], {{}});
                }} else if (typeof Jupyter !== 'undefined' && Jupyter.notebook && Jupyter.notebook.kernel) {{
                    Jupyter.notebook.kernel.execute(code);
                }} else {{
                    // Fallback: mostrar informações em alert
                    alert(`📊 Relatório de Disponibilidade: ${{subsystem}}\\n\\n` +
                          `Para ver o relatório completo, execute:\\n\\n` +
                          `show_availability_report("${{subsystem}}")\\n\\n` +
                          `Isso exibirá:\\n` +
                          `📈 Gráfico de disponibilidade dos últimos 12 meses\\n` +
                          `📋 Relatório detalhado com MTTF e MTTR\\n` +
                          `📊 Métricas de performance\\n` +
                          `💡 Resumo executivo com recomendações`);
                }}
            }}

            // Inicializar quando a página carregar
            document.addEventListener('DOMContentLoaded', function() {{
                console.log('DOM carregado, aguardando Chart.js...');
                waitForChartJS(function() {{
                    console.log('Chart.js disponível, inicializando gráficos...');
                    initializeCharts();
                }});
            }});

            // Fallback para garantir inicialização
            setTimeout(function() {{
                if (Object.keys(charts).length === 0) {{
                    console.log('Tentativa de inicialização de fallback...');
                    waitForChartJS(initializeCharts);
                }}
            }}, 3000);
        </script>
    </body>
    </html>
    '''

    return html_code

# Função para exibir a interface
def show_interface():
    """
    Exibe a interface HTML adaptada para Google Colab.
    """
    html_code = create_html_interface_for_colab()
    display(HTML(html_code))
```
