```python
data.columns = data.columns.str.strip()

data['data_abertura'] = pd.to_datetime(data['data_abertura'], format="%d/%m/%Y", errors='coerce')
data['Dt. Falha'] = pd.to_datetime(data['Dt. Falha'], format="%d/%m/%Y", errors='coerce')
data['data_fechamento'] = pd.to_datetime(data['data_fechamento'], format="%d/%m/%Y", errors='coerce')

# Limpar espaços e quebras de linha
data['Hr. Abertura'] = data['Hr. Abertura'].astype(str).str.strip().str.replace(r'\s+', '', regex=True)
data['Hr. Falha'] = data['Hr. Falha'].astype(str).str.strip().str.replace(r'\s+', '', regex=True)
data['Hora Enc.'] = data['Hora Enc.'].astype(str).str.strip().str.replace(r'\s+', '', regex=True)

# Converter horas (opcional, se quiser manter como string pode pular)
data['Hr. Abertura'] = pd.to_datetime(data['Hr. Abertura'], format="%H:%M", errors='coerce').dt.time
data['Hr. Falha'] = pd.to_datetime(data['Hr. Falha'], format="%H:%M", errors='coerce').dt.time
data['Hora Enc.'] = pd.to_datetime(data['Hora Enc.'], format="%H:%M", errors='coerce').dt.time

# Converter colunas numéricas
data['Contador'] = pd.to_numeric(data['Contador'], errors='coerce')

# Outras colunas de texto podem continuar como string
texto_cols = [
    'Solicitacao', 'Bem/Localiz.', 'Desc.Bem/Loc', 'Centro Trab.', 'Nome Locali.',
    'Ramal', 'Sublocal', 'Situacao S.S', 'Prioridade', 'Descricao',
    'Solucao SS', 'Ordem Servic', 'Solicitante', 'Reclamante', 'Executante',
    'Nome Exec.', 'Tipo de SS', 'Ret.Operacao', 'Vandalismo', 'Criticidade', 'Tipo Solicit'
]
data[texto_cols] = data[texto_cols].astype("string")

# Conferir tipos
print(data.dtypes)
```

```python
from langchain_ollama import OllamaLLM  # se estiver usando LangChain + Ollama

# Crie o LLM informando o modelo
client = OllamaLLM(model="mistral")  # ou "mistral-7b" dependendo do que você baixou

def testing_question(question):
    """
    Summarizes the text using Mistral via Ollama.
    """
    summary = client.predict(f"You are a helpful assistant that summarizes text.\n\nUser: {question}\nAssistant:")
    return summary

# Teste
print(testing_question("Write a story about BTC"))
```

```python
def query_data_mistral(data_json, question, model_name):
    prompt = f"Based on the following relational data, answer the question:\n{data_json}\nQuestion: {question}"
    
    # No Ollama, usamos client.chat()
    chat_response = client.invoke(prompt)
    
    # Retorna a resposta gerada
    return chat_response

# Limitar os dados
limited_data = data.head(500)

# Converter DataFrame para JSON
data_json = limited_data.to_json(orient='records')

# Modelo Mistral no Ollama
model_name = 'mistral'  # ou 'mistral-7b', dependendo do que você baixou com `ollama pull`

# Pergunta de teste
question = "Summarize the text"

# Chamada
resposta = query_data_mistral(data_json, question, model_name)
print(resposta)
```

```python
import streamlit as st
import pandas as pd
import os
from PIL import Image

# Set Streamlit layout to wide
st.set_page_config(layout="wide")

# Load data from Excel
@st.cache_data
def load_data():
    df = pd.read_excel("D:\\Mestrado\\teste\\langchain_test\\data\\dados.xlsx")

    # Corrigir possíveis espaços/quebras de linha nos nomes das colunas
    df.columns = df.columns.str.strip()
    return df

data = load_data()

# ==============================
# CSS Customizado
# ==============================
st.markdown("""
    <style>
    body {
        background-color: #ffcccb;
    }
    
    .reportview-container .main .block-container {
        background-color: #ffcccb;  /* Light pink background for the main container */
    }

    .sidebar .sidebar-content {
        background-color: #ffcccb;  /* Light pink background for the sidebar */
    }

    .custom-card {
        background-color: white;
        padding: 5px;
        border-radius: 10px;
        border: 2px solid #d3d3d3;
        box-shadow: 2px 2px 5px rgba(0, 0, 0, 0.1);
        text-align: center;
        margin-bottom: 10px;
    }

    .custom-card h3 {
        margin-bottom: 2px;
    }

    .custom-card p {
        color: #003366;
        font-size: 20px;
        font-weight: bold;
    }

    .card-container {
        display: flex;
        justify-content: space-between;
        margin-bottom: 10px;
        gap: 10px;
    }

    .centered-title {
        text-align: center;
        font-size: 32px;
        font-weight: bold;
        margin-bottom: 20px;
    }

    .stDataFrame div[data-testid="stHorizontalBlock"] {
        width: auto !important;
        min-width: 150px !important;
    }
    </style>
    """, unsafe_allow_html=True)

# ==============================
# Imagens
# ==============================
image = Image.open('AI Pic3.png')
st.image(image, use_column_width=True, caption="Sales and AI Banner", output_format="PNG")

# Título centralizado
st.markdown('<h1 class="centered-title">FerrovIA Dashboard</h1>', unsafe_allow_html=True)

# Sidebar com imagem
image2 = Image.open('AI Pic1.png')
st.sidebar.image(image2, use_column_width=True)

# ==============================
# Filtros na Sidebar
# ==============================
st.sidebar.header("Filter Options")

hotel_options = data["Bem/Localiz."].dropna().unique() if "Bem/Localiz." in data else []
year_options = data["data_abertura"].dropna().unique() if "data_abertura" in data else []
month_options = data["Hr. Abertura"].dropna().unique() if "Hr. Abertura" in data else []
market_segment_options = data["Prioridade"].dropna().unique() if "Prioridade" in data else []
country_options = data["Reclamante"].dropna().unique() if "Reclamante" in data else []

hotel_filter = st.sidebar.multiselect("Bem/Localiz.", hotel_options, default=list(hotel_options))
year_filter = st.sidebar.multiselect("data_abertura", year_options, default=list(year_options))
month_filter = st.sidebar.multiselect("Hr. Abertura", month_options, default=list(month_options))
market_segment_filter = st.sidebar.multiselect("Prioridade", market_segment_options, default=list(market_segment_options))
country_filter = st.sidebar.selectbox("Reclamante", ["All"] + list(country_options))

# Data Limit filter
data_limit = st.sidebar.slider("Limit Data (Rows)", min_value=10, max_value=120000, value=200)

# ==============================
# Aplicando Filtros
# ==============================
filtered_data = data.copy()

if "Bem/Localiz." in filtered_data:
    filtered_data = filtered_data[filtered_data["Bem/Localiz."].isin(hotel_filter)]

if "data_abertura" in filtered_data:
    filtered_data = filtered_data[filtered_data["data_abertura"].isin(year_filter)]

if "Hr. Abertura" in filtered_data:
    filtered_data = filtered_data[filtered_data["Hr. Abertura"].isin(month_filter)]

if "Prioridade" in filtered_data:
    filtered_data = filtered_data[filtered_data["Prioridade"].isin(market_segment_filter)]

if country_filter != "All" and "Reclamante" in filtered_data:
    filtered_data = filtered_data[filtered_data["Reclamante"] == country_filter]

filtered_data = filtered_data.head(data_limit)

# ==============================
# Métricas
# ==============================
num_rows = len(filtered_data)
num_tokens = num_rows * 21 * 6
num_countries = filtered_data['Reclamante'].nunique() if "Reclamante" in filtered_data else 0
num_cancellations = filtered_data['Contador'].astype(float).sum() if "Contador" in filtered_data else 0

# Cards com métricas
st.markdown('<div class="card-container">', unsafe_allow_html=True)
col1, col2, col3, col4, col5 = st.columns(5)

with col1:
    st.markdown(f'<div class="custom-card"><h3>Number of Rows</h3><p>{num_rows:,}</p></div>', unsafe_allow_html=True)

with col2:
    st.markdown(f'<div class="custom-card"><h3>Est. of Tokens</h3><p>{num_tokens:,}</p></div>', unsafe_allow_html=True)

with col3:
    st.markdown(f'<div class="custom-card"><h3>Number of Countries</h3><p>{num_countries:,}</p></div>', unsafe_allow_html=True)

with col4:
    st.markdown(f'<div class="custom-card"><h3>No. of Cancellations</h3><p>{num_cancellations:,}</p></div>', unsafe_allow_html=True)

st.markdown('</div>', unsafe_allow_html=True)

# ==============================
# Função para consultar modelo
# ==============================
# ATENÇÃO: 'client' e 'model_name' devem estar definidos antes

def query_data(data_json, question):
    prompt = f"Based on the following relational data, answer the question:\n{data_json}\nQuestion: {question}"
    chat_completion = client.invoke(prompt, model=model_name)
    return chat_completion.choices[0].message.content

# ==============================
# Layout com Perguntas e Tabela
# ==============================
col_left, col_right = st.columns(2)

# Perguntas (esquerda)
with col_left:
    st.subheader("Ask a Question About the Data")
    user_question = st.text_input("Enter your question:")

    if st.button("Ask AI"):
        if user_question:
            answer = query_data(filtered_data.to_json(orient="records"), user_question)
            st.subheader("AI Answer")
            st.write(answer)
        else:
            st.warning("Please enter a question.")

# Tabela (direita)
with col_right:
    st.subheader("Filtered Hotel Bookings Data")
    st.dataframe(filtered_data.astype(str))

```
